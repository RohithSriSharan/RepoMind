 `__name__ ="__main__"`

`"I use if __name__ == '__main__': as a guard so the code inside it only runs when this module is executed directly — not when it's imported by another module. Python automatically sets a special variable, __name__, to the string '__main__' when a file is run directly, and to the module's own import path (like 'src.logging_config') when it's imported elsewhere. So the condition is True only when you run the file directly, and False when it's imported — which prevents test/demo code from accidentally executing every time some other file imports functions from this one. In my project, I use it for a quick smoke test — a minimal check that the module's basic functionality works, without running a full test suite."`
