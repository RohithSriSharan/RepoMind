# PART 1 — LISTS

## 1. What is a list?
- An **ordered**, **mutable** (changeable) collection of items.
- Can hold mixed types: `[1, "two", 3.0, True, [1,2]]`
- Defined with square brackets: `[]`

```python
nums = [1, 2, 3, 4, 5]
empty = []
mixed = [1, "two", 3.0, True]
```

## 2. Indexing & Slicing

```python
nums = [10, 20, 30, 40, 50]

nums[0]      # 10   (first element)
nums[-1]     # 50   (last element)
nums[2]      # 30   (index 2)

# Slicing: lst[start:stop:step]
nums[1:3]    # [20, 30]
nums[:3]     # [10, 20, 30]
nums[2:]     # [30, 40, 50]
nums[:]      # [10,20,30,40,50] - full COPY
nums[::-1]   # [50,40,30,20,10] - reversed
nums[::2]    # [10, 30, 50]
nums[::-2]   # [50, 30, 10]
```

**Out-of-range slicing does NOT crash**, but out-of-range indexing does:
```python
nums[2:100]   # fine, returns what exists
nums[100]     # IndexError
```

## 3. Adding elements

| Method | What it does |
|---|---|
| `lst.append(x)` | adds `x` to the **end** |
| `lst.insert(i, x)` | inserts `x` at index `i` |
| `lst.extend([a,b])` | adds multiple items to the end |
| `lst += [a,b]` | same as `.extend()` |
| `lst + [a,b]` | returns a **new** list (doesn't mutate `lst`) |

```python
a = [1]
a.append([2,3])   # [1, [2, 3]]  <- nested!
b = [1]
b.extend([2,3])   # [1, 2, 3]     <- flat
```

## 4. Removing elements

| Method | Removes by | Returns removed item? |
|---|---|---|
| `lst.remove(x)` | **value** (first match) | No |
| `lst.pop()` | last item | Yes |
| `lst.pop(i)` | **index** `i` | Yes |
| `del lst[i]` | index `i` | No |
| `del lst[i:j]` | slice range | No |
| `lst.clear()` | everything | No |

⚠️ `.remove(x)` = by *value*. `.pop(i)`/`del lst[i]` = by *index*. Easy to mix up.

⚠️ **Don't mutate a list while iterating over it directly** — elements get skipped.
```python
# BUGGY
for i in lst:
    if i % 3 == 0:
        lst.remove(i)

# FIX 1 - loop over a copy
for i in lst[:]:
    if i % 3 == 0:
        lst.remove(i)

# FIX 2 - loop backwards by index
for i in range(len(lst)-1, -1, -1):
    if lst[i] % 3 == 0:
        lst.pop(i)
```

## 5. Sorting & reversing

```python
nums.sort()              # in-place, ascending, returns None
nums.sort(reverse=True)   # in-place, descending
nums.reverse()            # in-place

sorted(nums)               # returns a NEW sorted list
words.sort(key=len)         # custom key
words.sort(key=lambda w: w[-1])
```

⚠️ `sort()` returns `None` — never do `x = lst.sort()`.

## 6. Searching & info

```python
3 in nums          # membership check, by value
nums.index(20)     # index of first match (ValueError if missing)
nums.count(20)     # occurrences
len(nums)
```

## 7. Iteration patterns

```python
for x in nums: ...
for i, x in enumerate(nums): ...
for x in reversed(nums): ...
for a, b in zip(list1, list2): ...
```

## 8. List comprehensions

```python
squares = [x**2 for x in range(5)]
evens   = [x for x in nums if x % 2 == 0]
doubled = [x*2 for x in nums if x > 10]
nested  = [y for row in matrix for y in row]
combo   = [(x,y) for x in [1,2] for y in [3,4]]
```

## 9. Copying — critical gotcha

```python
a = [1, 2, 3]
b = a              # SAME object, not a copy
c = a.copy()        # real shallow copy
c = a[:]             # also shallow copy
import copy
d = copy.deepcopy(a) # needed for nested lists/dicts inside
```

```python
nested = [[1,2], [3,4]]
shallow = nested.copy()
shallow[0].append(99)
print(nested)   # [[1, 2, 99], [3, 4]]  <- inner list still shared!
```

## 10. Modifying via slicing

```python
nums[1:3] = []             # delete a range
nums[1:1] = [100, 200]      # insert without removing
nums[:] = [9,9,9]            # replace all contents in place
```

## 11. Nested lists

```python
matrix = [[1,2,3], [4,5,6], [7,8,9]]
matrix[1][2]     # 6
flat = [val for row in matrix for val in row]
```

## 12. Useful built-ins

```python
len(lst); max(lst); min(lst); sum(lst)
any(lst); all(lst)
list(range(5)); list("abc")
```

## 13. Common recipes

```python
# dedup, preserve order
def dedup(lst):
    seen = []
    for x in lst:
        if x not in seen:
            seen.append(x)
    return seen

# dedup, order doesn't matter
list(set(lst))

# two-pointer swap
left, right = 0, len(lst) - 1
while left < right:
    lst[left], lst[right] = lst[right], lst[left]
    left += 1; right -= 1

# merge two sorted lists
def merge_sorted(a, b):
    result, i, j = [], 0, 0
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            result.append(a[i]); i += 1
        else:
            result.append(b[j]); j += 1
    result.extend(a[i:]); result.extend(b[j:])
    return result

# swap without temp
a, b = b, a
lst[i], lst[j] = lst[j], lst[i]
```

## 14. Quick gotcha checklist
- `sort()` returns `None`.
- `.remove(x)` = value, `.pop(i)`/`del lst[i]` = index.
- Never mutate while iterating directly — loop over a copy or go backwards.
- `a = b` doesn't copy — use `.copy()`, `[:]`, or `deepcopy()`.
- Shallow copies don't protect nested lists — use `deepcopy` when nested.
- Slicing never raises `IndexError`; direct indexing does.

# PART 2 — TUPLES

## 1. What is a tuple?
- An **ordered**, **immutable** (cannot be changed after creation) collection.
- Defined with parentheses `()` — though the **comma** is what actually makes it a tuple.

```python
t = (1, 2, 3)
single = (5,)        # comma required! (5) alone is just an int in parens
no_parens = 1, 2, 3   # also a valid tuple
empty = ()
```

⚠️ **Classic trap:** `(5)` is an `int`, not a tuple. You need the trailing comma: `(5,)`.

## 2. Indexing & Slicing — same as lists (read-only)

```python
t = (10, 20, 30, 40)
t[0]        # 10
t[-1]       # 40
t[1:3]      # (20, 30)
t[::-1]     # (40, 30, 20, 10)
```

## 3. Immutability

```python
t = (1, 2, 3)
t[0] = 99        # TypeError: 'tuple' object does not support item assignment
```

- You **cannot** add, remove, or reassign elements.
- To "modify" a tuple, you convert it to a list, change it, then convert back:
```python
t = (1, 2, 3)
lst = list(t)
lst.append(4)
t = tuple(lst)     # (1, 2, 3, 4)
```

⚠️ **Nested mutable objects inside a tuple ARE still mutable:**
```python
t = ([1, 2], 3)
t[0].append(99)     # legal! t is now ([1, 2, 99], 3)
```
The tuple itself can't be reassigned/resized, but a *mutable object it contains* can still be changed in place.

## 4. Why use a tuple over a list?
- Signals "this data shouldn't change" (intent/safety).
- **Hashable** (if contents are hashable) → can be used as a **dict key** or put in a **set**. Lists cannot.
- Slightly faster and more memory-efficient than lists.

```python
locations = {}
locations[(17.4, 78.4)] = "Hyderabad"   # tuple as dict key — works
locations[[17.4, 78.4]] = "Hyderabad"   # TypeError: unhashable type: 'list'
```

## 5. Unpacking

```python
point = (3, 4)
x, y = point            # x=3, y=4

a, b, c = 1, 2, 3         # tuple packing/unpacking without parens

first, *rest = (1, 2, 3, 4)     # first=1, rest=[2,3,4]
*rest, last = (1, 2, 3, 4)       # rest=[1,2,3], last=4
a, *middle, b = (1, 2, 3, 4, 5)  # a=1, middle=[2,3,4], b=5

# swap in one line
a, b = b, a
```

## 6. Methods (only two! — reflects immutability)

```python
t = (1, 2, 2, 3)
t.count(2)     # 2 — occurrences of value
t.index(3)     # 2 — index of first match (ValueError if missing)
```

## 7. Iteration & built-ins — same as lists

```python
for x in t: ...
for i, x in enumerate(t): ...
len(t); max(t); min(t); sum(t)
3 in t
```

## 8. Tuple of tuples / records pattern

```python
students = [("Asha", 88), ("Raj", 92), ("Kim", 79)]

for name, score in students:
    print(name, score)

students.sort(key=lambda s: s[1], reverse=True)   # sort by score
```

## 9. `namedtuple` — tuples with named fields

```python
from collections import namedtuple

Point = namedtuple("Point", ["x", "y"])
p = Point(3, 4)
p.x, p.y        # 3, 4  (also still works as p[0], p[1])
```
Useful when a plain tuple's meaning ("what is index 0 vs index 1?") gets confusing.

## 10. Tuple vs List — quick comparison

| | List | Tuple |
|---|---|---|
| Mutable | Yes | No |
| Syntax | `[1,2,3]` | `(1,2,3)` |
| Hashable (usable as dict key/set item) | No | Yes (if contents hashable) |
| Methods | many (append, sort, remove...) | only `count`, `index` |
| Typical use | data that changes | fixed records, dict keys, function returns |

## 11. Quick gotcha checklist
- `(5)` is not a tuple — need `(5,)`.
- Tuples are immutable, but mutable objects *inside* them can still change.
- Tuples (not lists) can be dict keys / set members, because they're hashable.
- Parentheses are optional for tuple creation — the comma is what matters.
- Only `.count()` and `.index()` exist as methods.

# PART 3 — DICTIONARIES

## 1. What is a dict?
- An **unordered-by-concept-but-insertion-ordered-in-practice** (Python 3.7+ preserves insertion order), **mutable** collection of **key → value** pairs.
- Keys must be **unique** and **hashable** (strings, numbers, tuples of hashables). Values can be anything.

```python
student = {"name": "Asha", "age": 21, "scores": [88, 92]}
empty = {}
d = dict(name="Asha", age=21)     # alt constructor
```

## 2. Accessing values

```python
student["name"]              # "Asha" — KeyError if missing
student.get("name")           # "Asha" — None if missing (safe)
student.get("gpa", 0.0)        # 0.0 — custom default if missing
```

⚠️ **Use `.get()` when a key might not exist** — avoids `KeyError` crashes.

## 3. Adding / updating

```python
student["gpa"] = 3.8              # add new key, or overwrite if exists
student.update({"age": 22, "city": "Hyderabad"})  # bulk add/update
student.setdefault("clubs", []).append("Chess")     # add key only if missing, then use it
```

## 4. Removing

| Method | Behavior |
|---|---|
| `del d[key]` | removes key; `KeyError` if missing |
| `d.pop(key)` | removes & returns value; `KeyError` if missing (unless default given) |
| `d.pop(key, default)` | removes & returns value, or `default` if missing — no crash |
| `d.popitem()` | removes & returns the **last inserted** key-value pair |
| `d.clear()` | empties the whole dict |

```python
age = student.pop("age", None)   # safe removal
```

## 5. Checking membership

```python
"name" in student          # checks KEYS by default -> True
"Asha" in student.values()  # checks VALUES -> True
"name" in student.keys()     # same as "name" in student
```

⚠️ `in` on a dict checks **keys**, not values, by default. Common source of confusion.

## 6. Iteration patterns

```python
for key in student: ...                     # iterates KEYS by default
for key in student.keys(): ...
for value in student.values(): ...
for key, value in student.items(): ...       # most common — both at once
```

⚠️ **Don't add/remove keys while iterating over the dict directly** — raises `RuntimeError: dictionary changed size during iteration`. Loop over `list(d.keys())` or `list(d.items())` instead if you need to modify.

## 7. Dict comprehensions

```python
squares = {x: x**2 for x in range(5)}                  # {0:0, 1:1, 2:4, ...}
filtered = {k: v for k, v in student.items() if v}       # drop falsy values
inverted = {v: k for k, v in d.items()}                    # swap keys/values
```

## 8. Nested dicts

```python
company = {
    "engineering": {"Asha": 90000, "Raj": 85000},
    "sales": {"Kim": 70000}
}

company["engineering"]["Asha"]     # 90000

for dept, employees in company.items():
    for name, salary in employees.items():
        print(dept, name, salary)
```

## 9. Merging dictionaries

```python
a = {"x": 1, "y": 2}
b = {"y": 99, "z": 3}

merged = {**a, **b}          # {'x':1, 'y':99, 'z':3} — b wins on conflicts
a.update(b)                    # mutates a in place, same conflict rule
merged2 = a | b                 # Python 3.9+ merge operator
a |= b                           # Python 3.9+ in-place merge
```

## 10. Sorting a dict

Dicts don't sort in place — you sort the *items* and rebuild if needed:

```python
scores = {"Asha": 88, "Raj": 92, "Kim": 79}

sorted(scores.items(), key=lambda kv: kv[1], reverse=True)
# [('Raj', 92), ('Asha', 88), ('Kim', 79)]

dict(sorted(scores.items(), key=lambda kv: kv[1]))   # back into a dict, sorted by value
```

## 11. Counting pattern (very common interview pattern)

```python
words = ["a", "b", "a", "c", "b", "a"]

counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1
# {'a': 3, 'b': 2, 'c': 1}

# or, using collections
from collections import Counter
Counter(words)   # Counter({'a': 3, 'b': 2, 'c': 1})
```

## 12. `defaultdict` — avoids `.get()`/`.setdefault()` boilerplate

```python
from collections import defaultdict

groups = defaultdict(list)
for name, dept in [("Asha","eng"), ("Raj","eng"), ("Kim","sales")]:
    groups[dept].append(name)
# {'eng': ['Asha', 'Raj'], 'sales': ['Kim']}
```

## 13. Copying — same shallow/deep trap as lists

```python
a = {"x": [1, 2]}
b = a.copy()          # shallow copy
b["x"].append(99)
print(a)                # {'x': [1, 2, 99]}  <- inner list still shared!

import copy
c = copy.deepcopy(a)     # fully independent
```

## 14. Quick gotcha checklist
- `d[key]` crashes on missing key; `.get(key, default)` doesn't.
- `in` checks keys by default, not values.
- Keys must be hashable — lists can't be keys, tuples can.
- Don't resize a dict while iterating it directly.
- `.pop(key)` without a default crashes if key is missing.
- Shallow copy doesn't protect nested mutable values.
- Dicts preserve insertion order (Python 3.7+), but that's not the same as being "sorted."

# PART 4 — SETS

## 1. What is a set?
- An **unordered**, **mutable** collection of **unique**, **hashable** items — no duplicates, no indexing.
- Defined with curly braces `{}` or `set()`.

```python
s = {1, 2, 3}
s2 = set([1, 2, 2, 3])   # {1, 2, 3} — duplicates auto-removed
empty = set()              # NOT {} — that creates an empty DICT
```

⚠️ **`{}` is an empty dict, not an empty set.** You must use `set()` for an empty set.

## 2. No indexing, no order

```python
s = {3, 1, 2}
s[0]        # TypeError — sets aren't indexable
```
Sets don't preserve insertion order the way dicts/lists do (implementation detail: order depends on hashing) — never rely on set order.

## 3. Adding / removing

| Method | Behavior |
|---|---|
| `s.add(x)` | adds a single item |
| `s.update([a,b])` | adds multiple items from an iterable |
| `s.remove(x)` | removes item; `KeyError` if missing |
| `s.discard(x)` | removes item; **no error** if missing |
| `s.pop()` | removes & returns an **arbitrary** item |
| `s.clear()` | empties the set |

⚠️ Use `.discard()` when you're not sure the item exists — `.remove()` crashes on a miss, `.discard()` doesn't.

## 4. Set algebra (the main reason to use sets)

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

a | b     # union: {1,2,3,4,5,6}          — also a.union(b)
a & b     # intersection: {3,4}            — also a.intersection(b)
a - b     # difference: {1,2}               — items in a but not b — also a.difference(b)
a ^ b     # symmetric difference: {1,2,5,6}  — in one or the other, not both — also a.symmetric_difference(b)
```

## 5. Set comparisons

```python
a = {1, 2}
b = {1, 2, 3}

a.issubset(b)      # True — a <= b also works
b.issuperset(a)     # True — b >= a also works
a.isdisjoint({9,10}) # True — no overlap at all
a == {2, 1}           # True — order doesn't matter for equality
```

## 6. Membership & info

```python
3 in a          # O(1) fast lookup — this is a KEY reason to use sets
len(a)
```

⚠️ **Membership checks (`in`) are much faster in a set than in a list** for large collections — `O(1)` average vs `O(n)`. This is the #1 practical reason to convert a list to a set for lookups.

## 7. Iteration & comprehensions

```python
for x in a: ...              # order not guaranteed

evens = {x for x in range(10) if x % 2 == 0}   # set comprehension
```

## 8. Common uses

```python
# dedup a list (order NOT preserved)
list(set([1, 2, 2, 3, 1]))    # [1, 2, 3] (order may vary)

# fast membership testing
seen = set()
for item in big_list:
    if item in seen:
        print("duplicate:", item)
    seen.add(item)

# unique characters check
def has_unique_chars(s):
    return len(set(s)) == len(s)
```

## 9. `frozenset` — the immutable, hashable version of a set

```python
fs = frozenset([1, 2, 3])
fs.add(4)    # AttributeError — frozensets can't be modified

# because it's hashable, it CAN be used as a dict key or inside another set
d = {frozenset([1,2]): "pair A"}
```

## 10. Quick gotcha checklist
- `{}` creates a dict, not a set — use `set()` for an empty set.
- Sets have no guaranteed order — never index or rely on order.
- `.remove(x)` crashes if missing; `.discard(x)` doesn't.
- Set items must be hashable — no lists/dicts inside a set (tuples are fine).
- `in` on a set is fast (`O(1)`) — great for large-scale membership checks.
- `frozenset` = immutable set, usable as a dict key or inside another set.

## 1. Arrays & Hashing

**Core idea:** Use a hashmap/set for O(1) lookups to avoid nested loops (O(n²) → O(n)). Also covers grouping items by a derived "key."

**How to recognize it:**
- "Have I seen this before?" / duplicate detection
- Need to count frequency of things
- Need to find complements/pairs quickly
- Need to group items sharing some property

**Core template:**
```python
seen = {}  # or set()
for item in items:
    # check/update seen based on item
```

**5 most commonly asked questions:**
1. **Two Sum** — hashmap storing value→index, check for complement
2. **Contains Duplicate** — set to track seen values
3. **Valid Anagram** — frequency count comparison
4. **Group Anagrams** — sorted-string as key, `defaultdict(list)`
5. **Top K Frequent Elements** — count + sort/heap/bucket sort

**Other frequently asked:** Product of Array Except Self, Longest Consecutive Sequence, Valid Sudoku, Encode and Decode Strings.

---

## 2. Two Pointers

**Core idea:** Two pointers moving toward each other (converging) to avoid brute-force nested loops. Requires sorted input for the sum-comparison variant.

**How to recognize it:**
- Sorted array + need to find pair/triplet matching a condition
- Palindrome checking
- "Container"/area-maximizing problems

**Core template:**
```python
left, right = 0, len(arr) - 1
while left < right:
    if condition_met:
        # do something, move one or both pointers
    elif need_bigger:
        left += 1
    else:
        right -= 1
```

**5 most commonly asked questions:**
1. **Valid Palindrome** — skip non-alphanumeric, compare chars from both ends
2. **Two Sum II (sorted array)** — classic converging pointers
3. **3Sum** — fix one element + two pointers on the rest, skip duplicates
4. **Container With Most Water** — move the pointer at the shorter wall
5. **Trapping Rain Water** — track left_max/right_max, move smaller-max side

**Other frequently asked:** Sort Colors (Dutch national flag — 3 pointers), Remove Duplicates from Sorted Array (same-direction variant), Merge Sorted Array.

---

## 3. Sliding Window

**Core idea:** Maintain a "window" `[left, right]` over an array/string; expand `right`, shrink `left` when a condition is violated. Two flavors: **variable-size** (grows/shrinks based on validity) and **fixed-size** (window length is constant).

**How to recognize it:**
- "Longest/shortest substring/subarray that satisfies X"
- "At most k..." or "exactly k..." constraints
- Contiguous subarray/substring problems

**Core template (variable-size):**
```python
left = 0
for right in range(len(arr)):
    # expand: add arr[right] to window state
    while window_invalid:
        # shrink: remove arr[left] from window state
        left += 1
    # update answer using current window
```

**5 most commonly asked questions:**
1. **Best Time to Buy and Sell Stock** — track running min + max profit (the "loose" example)
2. **Longest Substring Without Repeating Characters** — set-based window, shrink on duplicate
3. **Longest Repeating Character Replacement** — `window_size - max_freq <= k` validity check
4. **Permutation in String** — fixed-size window, compare frequency counts
5. **Minimum Window Substring** — expand until valid, then shrink to minimize, track `have`/`required`

**Other frequently asked:** Sliding Window Maximum (uses a deque), Max Consecutive Ones III, Fruit Into Baskets.

## 4. Binary Search

**Core idea:** Repeatedly halve the search space by comparing the middle element to your target. Requires the search space to be sorted (or monotonic — including the "search on answer" variant, where you binary search over a range of possible answers rather than array indices).

**How to recognize it:**
- Array is sorted (or can be treated as sorted/monotonic)
- "Find target in O(log n)"
- "Find minimum/maximum value such that condition X holds" (search on answer space)
- Rotated sorted arrays
- Any problem where you can ask "is this guess too big or too small?" and get a clear yes/no

**Core template:**
```python
left, right = 0, len(arr) - 1
while left <= right:
    mid = (left + right) // 2
    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        left = mid + 1
    else:
        right = mid - 1
return -1  # not found
```

**5 most commonly asked questions:**
1. **Binary Search** — the classic template above, find target's index in sorted array
2. **Search Insert Position** — same as above, but return the index where target *would* be inserted if not found
3. **Search in Rotated Sorted Array** — sorted array rotated at an unknown pivot; determine which half is "properly sorted" at each step, then decide which half to search
4. **Find Minimum in Rotated Sorted Array** — similar rotation logic, but hunting for the pivot point itself
5. **Koko Eating Bananas** — "search on answer" variant: binary search over possible *eating speeds* (not array indices), checking if a given speed is "fast enough"

**Other frequently asked:** Find First and Last Position of Element in Sorted Array, Search a 2D Matrix, Time Based Key-Value Store, Median of Two Sorted Arrays (hard).

````python
def search(nums: list[int], target: int) -> int:
    left,right = 0 , len(nums) - 1

    while left <= right:
        mid = ( left + right )  // 2
        print(left, right ,mid)

        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid + 1
        elif nums[mid] > target:
            right = mid - 1
    return -1

            

print(search([-1,0,3,5,9,12], 9))   # Expected: 4
print(search([-1,0,3,5,9,12], 2))   # Expected: -1
````

````python
def searchInsert(nums: list[int], target: int) -> int:
    left , right = 0 , len(nums) -1 

    while left <= right :
        mid = (left + right ) // 2

        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid +1

        elif nums[mid] > target:
            right = mid -  1

    return left
        
    

print(searchInsert([1,3,5,6], 5))   # Expected: 2
print(searchInsert([1,3,5,6], 2))   # Expected: 1
print(searchInsert([1,3,5,6], 7))   # Expected: 4
````

````python
def isPerfectSquare(num: int) -> bool:
    left, right  = 0 , num
    while left <= right:
        mid = (left + right ) // 2
        square = mid * mid
        if square == num:
            return True
        elif square < num:

            left  = mid + 1
        elif square > num: 
            right = mid - 1

    return False



print(isPerfectSquare(16))   # Expected: True
print(isPerfectSquare(14))   # Expected: False
````

````python
def mySqrt(x: int) -> int:
    left , right = 0 , x

    while left <= right: 
        mid = ( left + right ) // 2
        square = mid * mid
        if square == x:
            return mid
        elif square >= x:
            right = mid - 1

        elif square <= x:
            left = mid + 1
      
    return right 

print(mySqrt(4))    # Expected: 2
print(mySqrt(8))    # Expected: 2 (floor of 2.828)
````

````python
def searchRange(nums: list[int], target: int) -> list[int]:
    left,right = 0, len(nums) - 1
    left_most = -1

    while left <= right:

        mid = (left + right) // 2
        if nums[mid] == target:
            left_most = mid
            right = mid - 1
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return left_most
                
                




print(searchRange([5,7,7,8,8,10], 8))   # Expected: [3, 4]
print(searchRange([5,7,7,8,8,10], 6))   # Expected: [-1, -1]
````
